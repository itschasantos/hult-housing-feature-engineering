```python
# Importing the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Loading the Data
feature_rich_data = pd.read_excel('./datasets/housing_feature_rich.xlsx')
```

```python
## Feature 1 ##

print("""
Feature Engineering - Mean Score for Ames Housing Dataset

Purpose:
- To create a 'Mean Score' that encapsulates a holistic view of each house's attributes 
    and identify whether the standardized mean of attributes matter when considering buying a house.
- This score combines various aspects such as size, quality, condition, and amenities into a single metric.

Method:
- Selected a range of features representing different aspects of a house.
- Standardized continuous and ordinal features to ensure equal contribution to the score regardless of their original scales.
- Calculated the score as the mean of these features, providing a balanced summary of each house.
""")

# Defining a list of features to include in the mean score
# These features are selected to provide a comprehensive view of each house, 
# covering aspects like size, quality, amenities, and condition.
features_for_checklist = [
    'Total_Bsmt_SF', 'Mas_Vnr_Area','Gr_Liv_Area', 'Garage_Area', 'Lot_Area', 'First_Flr_SF', 'Second_Flr_SF',
    'Overall_Qual', 'Overall_Cond', 'Fireplaces', 'TotRms_AbvGr', 'Full_Bath', 'Half_Bath', 
    'Porch_Area' 
]

# Data needs to be standardized to be re-evaluated across all features selected
# Separating the features into continuous, ordinal, and count since each type needs different approaches
continuous_features = ['Total_Bsmt_SF', 'Gr_Liv_Area', 'Garage_Area', 'Lot_Area', 'First_Flr_SF', 'Second_Flr_SF', 'Porch_Area']
ordinal_features = ['Overall_Qual', 'Overall_Cond']
count_features = ['Fireplaces', 'TotRms_AbvGr', 'Full_Bath', 'Half_Bath']

# Creating 'Sale_Price_Log' by applying logarithmic transformation to 'Sale_Price'
feature_rich_data['Sale_Price_Log'] = np.log(feature_rich_data['Sale_Price'])

# Standardizing continuous and ordinal features to ensure each feature contributes equally to the final score, irrespective of their original scale or units.
# Using MinMaxScaler to recompute based on actual range of each feature that needs to be normalized
scaler = MinMaxScaler()
feature_rich_data[continuous_features + ordinal_features] = scaler.fit_transform(feature_rich_data[continuous_features + ordinal_features])

# Standardizing count features if their distribution justifies it
feature_rich_data[count_features] = scaler.fit_transform(feature_rich_data[count_features])

# Combining all selected features to compute the Mean_Score
all_features = continuous_features + ordinal_features + count_features

# Calculating the checklist score as the mean of the selected features This score aggregates various aspects of a house into a single metric, 
# facilitating a more holistic comparison between houses.
feature_rich_data['Mean_Score'] = feature_rich_data[all_features].mean(axis=1).round(decimals = 3)

# Creating Mean_Score_Log by applying Logarithim transformation to Mean_Score
feature_rich_data['Mean_Score_Log'] = np.log1p(feature_rich_data['Mean_Score']).round(decimals = 3)

# Calculate correlations
correlation_with_sale_price_mean_score = feature_rich_data['Mean_Score'].corr(feature_rich_data['Sale_Price']).round(decimals=3)
correlation_with_sale_price_log_mean_score = feature_rich_data['Mean_Score'].corr(feature_rich_data['Sale_Price_Log']).round(decimals=3)
correlation_with_sale_price_mean_score_log = feature_rich_data['Mean_Score_Log'].corr(feature_rich_data['Sale_Price']).round(decimals=3)
correlation_with_sale_price_log_mean_score_log = feature_rich_data['Mean_Score_Log'].corr(feature_rich_data['Sale_Price_Log']).round(decimals=3)

print("Correlation with Sale Price (Mean Score):", correlation_with_sale_price_mean_score)
print("Correlation with Log Sale Price (Mean Score):", correlation_with_sale_price_log_mean_score)
print("Correlation of Log Mean Score with Sale Price:", correlation_with_sale_price_mean_score_log)
print("Correlation of Log Mean Score with Log Sale Price:", correlation_with_sale_price_log_mean_score_log)

#Checking if transformations are needed

# Setting up the figure for a 2x2 grid of scatter plots
plt.figure(figsize=(12, 12))

# Scatter plot for the actual Sale_Price vs Mean Score
plt.subplot(2, 2, 1)  # First plot in the top left corner
sns.scatterplot(x='Mean_Score', y='Sale_Price', data=feature_rich_data)
plt.title('Sale Price vs Mean Score')
plt.xlabel('Mean Score')
plt.ylabel('Sale Price ($)')

# Scatter plot for the log-transformed Sale_Price vs Mean Score
plt.subplot(2, 2, 2)  # Second plot in the top right corner
sns.scatterplot(x='Mean_Score', y='log_Sale_Price', data=feature_rich_data)  # Assuming log_Sale_Price is correct
plt.title('Log of Sale Price vs Mean Score')
plt.xlabel('Mean Score')
plt.ylabel('Log of Sale Price')

# Applying log transformation to both Checklist_Score and Sale_Price
# Assuming this transformation has been applied

# Scatter plot for the Log-Transformed Mean Score vs Log-Transformed Sale Price
plt.subplot(2, 2, 3)  # Third plot in the bottom left corner
sns.scatterplot(x='Mean_Score_Log', y='Sale_Price_Log', data=feature_rich_data)
plt.title('Log-Transformed Sale Price vs Log-Transformed Mean Score')
plt.xlabel('Log-Transformed Mean Score')
plt.ylabel('Log-Transformed Sale Price')

# Scatter plot for Log-Transformed Mean Score vs Sale Price
plt.subplot(2, 2, 4)  # Fourth plot in the bottom right corner
sns.scatterplot(x='Mean_Score_Log', y='Sale_Price', data=feature_rich_data)  # Adjusting according to your structure
plt.title('Sale Price vs Log-Transformed Mean Score')
plt.xlabel('Log-Transformed Mean Score')
plt.ylabel('Sale Price ($)')

plt.tight_layout()  # Adjust the layout to make sure there's no overlap
plt.show()
```

```python
# Feature 2

print("""
Purpose:
- The purpose of developing this feature is to offer a metric that captures the average living space per room,
 excluding bathrooms. 
 
- This ratio aims to highlight the balance between the size of the living area and the number 
of rooms, providing a deeper insight into the functional utility and spatial efficiency of a house. 

- It reflects how space is allocated across rooms, influencing a buyer's decision based on their preference for larger 
communal areas versus the number of rooms.

Method:
- Created by dividing the total gross living area by the total number of rooms above ground, this feature quantifies the average living space available per room. 
It offers a straightforward metric to assess the spaciousness of a house.

- Applied a logarithmic transformation to stabilize the variance and normalize the distribution of the ratio. 
This step helps in mitigating the skewness caused by houses with exceptionally large or small living areas, making the feature more robust for statistical analysis.

- Evaluated the feature's relationship with the house's sale price through correlation analysis and visual representation. 
This examination sheds light on how variations in living space per room affect property value, aiding in understanding market preferences.

""")

feature_rich_data1 = pd.read_excel('./datasets/housing_feature_rich.xlsx')

# Creating the feature
feature_rich_data1['Gross_living_area_to_total_rooms_ratio'] = (feature_rich_data1['Gr_Liv_Area'] / feature_rich_data1['TotRms_AbvGr']).round(decimals = 3)

# Calculating log of the feature
feature_rich_data1['Gross_living_area_to_total_rooms_ratio_Log'] = np.log(feature_rich_data1['Gross_living_area_to_total_rooms_ratio'])

# Calculating Sale Price Log
feature_rich_data1['Sale_Price_Log'] = np.log(feature_rich_data1['Sale_Price'])

# Calculating correlations
correlation_with_sale_price = feature_rich_data1['Gross_living_area_to_total_rooms_ratio'].corr(feature_rich_data1['Sale_Price']).round(decimals = 3)
correlation_with_sale_price_log = feature_rich_data1['Gross_living_area_to_total_rooms_ratio'].corr(feature_rich_data1['Sale_Price_Log']).round(decimals = 3)
correlation_with_sale_price1 = feature_rich_data1['Gross_living_area_to_total_rooms_ratio_Log'].corr(feature_rich_data1['Sale_Price']).round(decimals = 3)
correlation_with_sale_price_log1 = feature_rich_data1['Gross_living_area_to_total_rooms_ratio_Log'].corr(feature_rich_data1['Sale_Price_Log']).round(decimals = 3)

print("Correlation with Sale Price:", correlation_with_sale_price)
print("Correlation with Log Sale Price:", correlation_with_sale_price_log)
print("Correlation of Log X-variable with Sale Price:", correlation_with_sale_price1)
print("Correlation with Log X-variable with Log Sale Price:", correlation_with_sale_price_log1)

# Setting up the figure for a 2x2 grid of scatter plots
plt.figure(figsize=(12, 12))

# Scatterplot with Sale Price
plt.subplot(2, 2, 1)  # First plot in the top left corner
sns.scatterplot(x='Gross_living_area_to_total_rooms_ratio', y='Sale_Price', data=feature_rich_data1)
plt.title('Avg. Room Size vs Sale Price')
plt.xlabel('Avg. Room Size (sqft)')
plt.ylabel('Sale Price ($)')

# Scatterplot with Log Sale Price
plt.subplot(2, 2, 2)  # Second plot in the top right corner
sns.scatterplot(x='Gross_living_area_to_total_rooms_ratio', y='Sale_Price_Log', data=feature_rich_data1)
plt.title('Avg. Room Size vs Log Sale Price')
plt.xlabel('Avg. Room Size (sqft)')
plt.ylabel('Sale Price (log)')

# Scatterplot Gross_living_area_to_total_rooms_ratio_Log with Sale Price
plt.subplot(2, 2, 3)  # Third plot in the bottom left corner
sns.scatterplot(x='Gross_living_area_to_total_rooms_ratio_Log', y='Sale_Price', data=feature_rich_data1)
plt.title('Log Avg. Room Size vs Sale Price')
plt.xlabel('Avg. Room Size (log)')
plt.ylabel('Sale Price (sqft)')

# Scatterplot Gross_living_area_to_total_rooms_ratio_Log with Log Sale Price
plt.subplot(2, 2, 4)  # Fourth plot in the bottom right corner
sns.scatterplot(x='Gross_living_area_to_total_rooms_ratio_Log', y='Sale_Price_Log', data=feature_rich_data1)
plt.title('Log Avg. Room Size vs Log Sale Price')
plt.xlabel('Avg. Room Size (log)')
plt.ylabel('Sale Price (log)')

plt.tight_layout()  # Adjust subplots to fit into the figure area.
plt.show()
```

```python
# Feature 3

print("""

Feature Engineering - Excess Garage Space 

Purpose:
- To quantify the additional utility space available in the garage beyond just parking cars.
- This feature captures the extra space that can be utilizedas a multiroom eg.workshop, gym, storage

Method:
- Calculate the excess space by subtracting the space needed for cars from the total garage area.
- Use a predefined average car size to estimate the required space for the number of cars.
""")

# Loading the datasets
ames_non_continuous = pd.read_excel('./datasets/ames_non_continuous.xlsx')
ames_transformed = pd.read_excel('./datasets/ames_transformed.xlsx')

# Merge the datasets on the 'property_id' column
merged_data = pd.merge(ames_non_continuous, ames_transformed, on='property_id')

# Assuming an average car size to calculate excess garage space
avg_car_size = 160  # average size in square feet per car

# Calculate Excess Garage Space
merged_data['Excess_Garage_Space'] = merged_data['Garage_Area'] - (merged_data['Garage_Cars'] * avg_car_size)

# Ensure there are no negative values for logarithmic transformation
merged_data['Excess_Garage_Space'] = merged_data['Excess_Garage_Space'].apply(lambda x: max(x, 1))

# Apply logarithmic transformation to 'Excess_Garage_Space'
merged_data['Excess_Garage_Space_Log'] = np.log(merged_data['Excess_Garage_Space'])

# Creating 'Sale_Price_Log' by applying logarithmic transformation to 'Sale_Price'
merged_data['Sale_Price_Log'] = np.log(merged_data['Sale_Price'])

# Calculate correlations
correlation_egs_sp = merged_data['Excess_Garage_Space'].corr(merged_data['Sale_Price']).round(decimals=3)
correlation_egs_lsp = merged_data['Excess_Garage_Space'].corr(merged_data['Sale_Price_Log']).round(decimals=3)
correlation_egsl_sp = merged_data['Excess_Garage_Space_Log'].corr(merged_data['Sale_Price']).round(decimals=3)
correlation_egsl_lsp = merged_data['Excess_Garage_Space_Log'].corr(merged_data['Sale_Price_Log']).round(decimals=3)

print("Correlation between Excess Garage Space and Sale Price:", correlation_egs_sp)
print("Correlation between Excess Garage Space and Log Sale Price:", correlation_egs_lsp)
print("Correlation between Log of Excess Garage Space and Sale Price:", correlation_egsl_sp)
print("Correlation between Log of Excess Garage Space and Log Sale Price:", correlation_egsl_lsp)

# Visualizing the relationships in a 2x2 grid
plt.figure(figsize=(12, 12))

# Excess Garage Space vs Sale Price
plt.subplot(2, 2, 1)
sns.scatterplot(x='Excess_Garage_Space', y='Sale_Price', data=merged_data)
plt.title('Sale Price vs Excess Garage Space')
plt.xlabel('Excess Garage Space (sqft)')
plt.ylabel('Sale Price ($)')

# Excess Garage Space vs Log Sale Price
plt.subplot(2, 2, 2)
sns.scatterplot(x='Excess_Garage_Space', y='Sale_Price_Log', data=merged_data)
plt.title('Log Sale Price vs Excess Garage Space')
plt.xlabel('Excess Garage Space (sqft)')
plt.ylabel('Log Sale Price')

# Log of Excess Garage Space vs Sale Price
plt.subplot(2, 2, 3)
sns.scatterplot(x='Excess_Garage_Space_Log', y='Sale_Price', data=merged_data)
plt.title('Sale Price vs Log of Excess Garage Space')
plt.xlabel('Log of Excess Garage Space')
plt.ylabel('Sale Price ($)')

# Log of Excess Garage Space vs Log Sale Price
plt.subplot(2, 2, 4)
sns.scatterplot(x='Excess_Garage_Space_Log', y='Sale_Price_Log', data=merged_data)
plt.title('Log Sale Price vs Log of Excess Garage Space')
plt.xlabel('Log of Excess Garage Space')
plt.ylabel('Log Sale Price')

plt.tight_layout()
plt.show()
```

After hypothesizing on new potential features that can affect sales prices for housing, the team has short-listed it to the following three features:  

**1.	Mean Score:**

Reviewing the correlation between Mean Score and Sales Price, we see that the correlation is significantly higher than the other two engineered features. Intuitively, this also makes sense as the Mean Score feature is based on 14 variables, whereas both Average Room Size and Excess Garage Space are based on two variables. The combination of all these variables makes the feature capable of absorbing inconsistencies in the relationship between Sales Price and each of the 14 variables because negative inconsistencies in one relationship can be canceled out by positive inconsistencies in another relationship. In addition, as we are comparing correlation and not R^2, we are not assessing the proportion of variance in Sales Price, which the engineered feature's variance can explain, nor are we penalizing the significantly larger amount of predictors being used in this feature via the adjusted R^2. 

Looking at the non-transformed scatterplot for the mean score, we see an interesting trend which, in hindsight, makes sense. The Sales Price is exponentially increasing as the Mean Score increases; this is expected to be due to the general pricing of quality in general. Going from low quality to average quality, you usually see a lower price increase than when you go from average quality to luxury, as you pay a premium for luxury. This is likely what is causing the non-linear relationship between the two features, as the Mean Score can be translated to the overall quality of the housing. 

Comparing the different combinations of transformations, we see that the double log transformation best makes a linear relationship between the dependent variable and the engineered feature. If used in future models, we need to be aware that the interpretation of the coefficient would be a 1% increase in Mean Score leads to an x% increase in sales price. 


**2.	Average Room Size:**

As expected, we see a positive correlation between the engineered feature Average Room Size and Sales Price. Looking at the non-transformed scatterplot, we surprisingly see the opposite relationship compared to what we hypothesized. Initially, the team thought that customers would value an increase in average room size at the bottom of the spectrum more than at the middle and end of the spectrum, resembling a logarithmic curve. However, what we see is closer to an exponential curve with a low sales price impact at the low end of the spectrum and an increasing impact after the 150-175 average. sqft mark. Furthermore, we see a significant increase in variation after the 225 avg. sqft mark. These two marks, 150 and 225, could possibly represent that everything under 150 does not give you additional possibilities to utilize the rooms (you will feel that the rooms are small), and everything after 225 does not add value in the same way, hence the increased variation.  

Comparing the different combinations of transformations, we see that the double log transformation is again best at making a linear relationship between the sales price and the engineered feature. Like with the previous feature, we need to consider this for interpretation if used in future models. 


**3.	Excess Garage space:**

The Excess Garage Space is the worst-performing feature in terms of correlation. Looking at the non-transformed scatterplot, we potentially find some of the causes of this. Firstly, we have many garages with no excess space yet differentiating Sales Prices. Secondly, we see a lot of stacked values, again with different Sales Prices, around the 225 sqft mark. Lastly, we see that after the 175 sqft mark, the variation increases a lot, indicating a less strong relationship after this point.  

This could indicate that garages have specific standardized sizes. Additionally, it could indicate that the feature should be further engineered to either an ordinal variable with little, middle and a lot of excess space based on ranges or into a binary variable indicating whether a garage has excess space. 

However, the correlation is still positive and 0.49, and comparing the different scatterplots, we see that the non-transformed feature is performing best. However, if used in conjunction with the two other features, forcing sales price to be logged, the correlation drops to 0.48.
